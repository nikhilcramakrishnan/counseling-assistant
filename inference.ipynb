```python
# Use a pipeline as a high-level helper
from transformers import pipeline
pipe = pipeline("text-generation", model="nikhil-c-r/counsel_assist")
```

```python
messages = [
    {"role": "user", "content": "I couldn't sleep for few days. Whenever i sleeps, bad dreams comes"},
]
```

```python
pipe(messages)
```
